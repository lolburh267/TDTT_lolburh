```python
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self,w,h):
        self.w=w
        self.h=h
    def area(self):
        return self.w*self.h
class Circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r*self.r
loai_hinh=input().strip()
tham_so=input().strip().split()
if loai_hinh=="RECTANGLE":
    w=float(tham_so[0])
    h=float(tham_so[1])
    hinh_chu_nhat=Rectangle(w,h)
    print(f"{hinh_chu_nhat.area():.2f}")
elif loai_hinh=="CIRCLE":
    r=float(tham_so[0])
    hinh_tron=Circle(r)
    print(f"{hinh_tron.area():.2f}")
```

# W8A17

```python
class Vehicle:
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed
    def info(self):
        return f"{self.brand}-{self.speed}"
class Car(Vehicle):
    def __init__(self,brand,speed,seats):
        super().__init__(brand,speed)
        self.seats=seats
    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"
loai_doi_tuong=input().strip()
du_lieu=input().strip().split()
if loai_doi_tuong=="Vehicle":
    obj=Vehicle(brand=du_lieu[0],speed=du_lieu[1])
    print(obj.info())
elif loai_doi_tuong=="Car":
    obj=Car(brand=du_lieu[0],speed=du_lieu[1],seats=du_lieu[2])
    print(obj,info())
```

# W8A18

```python
class Polygon:
    def perimeter(self):
        pass
class Triangle(Polygon):
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def perimeter(self):
        return self.a +self.b +self.c
a,b,c=map(int,input().split())
tam_giac=Triangle(a,b,c)
print(tam_giac.perimeter())
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def describe(self):
        return f"{self.name} {self.age}"
class Student(Person):
    def __init__(self,name,age,school):
        super().__init__(name,age)
        self.school=school
    def describe(self):
        return f"{self.name} {self.age} {self.school}"
loai_doi_tuong=input().strip()
du_lieu=input().strip().split()
if loai_doi_tuong=="Person":
    obj=Person(name=du_lieu[0],age=du_lieu[1])
    print(obj.describe())
elif loai_doi_tuong=="Student":
    obj=Student(name=du_lieu[0],age=du_lieu[1],school=du_lieu[2])
    print(obj.describe)
```

```python
class Account:
    def __init__(self,balance):
        self.balance=balance
    def deposit(self,x):
        self.balance+=x
    def get_balance(self):
        return self.balance
class SavingAccount(Account):
    def __init__(self,balance,rate):
        super().__init__(balance)
        self.rate=rate
    def add_interest(self):
        self.balance+=self.balance*self.rate
ip=input().strip().split()
if len(ip)==2:
    balance=float(ip[0])
    rate=float(ip[1])
    my_account=SavingAccount(balance,rate)
    my_account.add_interest()
    so_du_moi=my_account.get_balance()
    print(f"{so_du_moi:.2f}")
```
